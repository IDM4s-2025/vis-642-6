```python
import pandas as pd
import requests

# Fetch the data.
df = pd.read_csv("https://ourworldindata.org/grapher/gdp-per-capita-worldbank.csv?v=1&csvType=full&useColumnShortNames=true", storage_options = {'User-Agent': 'Our World In Data data fetch/1.0'})

# Fetch the metadata
metadata = requests.get("https://ourworldindata.org/grapher/gdp-per-capita-worldbank.metadata.json?v=1&csvType=full&useColumnShortNames=true").json()
```

```python
df.head()
```

```python
df_2 = df[df["Year"] == 2023].reset_index(drop=True)
df_2 = df_2[["Code", "ny_gdp_pcap_pp_kd"]]
df_2 = df_2.rename(columns={"Code": "LOCATION"})

df_2.head(20)
```

```python
df_tabla = pd.read_csv("OECD,DF_BLI,+all.csv")
df_tabla.head(5)
```

```python
df_tabla2 = df_tabla[["LOCATION","INDICATOR","OBS_VALUE"]]
df_tabla2.head(5)
```

```python

df_tabla3 = df_tabla2.groupby(["LOCATION", "INDICATOR"]).first().reset_index()

df_pivot = df_tabla3.pivot(index="LOCATION", columns="INDICATOR", values="OBS_VALUE")
df_pivot.head()
```

```python
df_final = df_pivot.merge(df_2[["LOCATION", "ny_gdp_pcap_pp_kd"]], on="LOCATION", how="inner")

df_final.head()
```

```python
df_final.to_csv("indicatos_final.csv", index=False)
```

```python
df_normalizados = pd.read_csv("indicadores_normalizados"".csv")
```

```python
df_normalizados.head()
```

```python
df_normalizados.info()
```

```python
df_normalizados["Mean"] = df_normalizados.iloc[:,1:].mean(axis=1)
```

```python
df_normalizados = df_normalizados.sort_values(by="Mean", ascending=False)
```

```python

df_normalizados.head()
```

```python
import plotly.express as px

top_n = 25
df_top = df_normalizados.head(top_n)

fig = px.bar(
    df_top,
    x="Mean",
    y=df_top.columns[0],
    orientation='h',
    color="Mean",  
    color_continuous_scale="Viridis",
    text="Mean", 
    title=f"Top {top_n} Países por Promedio de Indicadores"
)
fig.update_traces(texttemplate='%{text:.2f}', textposition='outside')
fig.update_layout(
    width=800,
    height=600,
    yaxis={'categoryorder':'total ascending'},  
    xaxis_title="Promedio de Indicadores",
    yaxis_title="País"
)
fig.show()
```

```python
import plotly.express as px
import numpy as np


corr = df_normalizados.iloc[:, 1:].corr()


import numpy as np
corr_filtered = corr.where((corr > 0.6) | (corr < -0.6))


fig = px.imshow(
    corr_filtered,
    labels=dict(x="Indicador", y="Indicador", color="Correlación"),
    title="Matriz de Correlación (>|0.6|) entre Indicadores",
    width=800,
    height=800,
    color_continuous_scale=px.colors.sequential.Turbo,
)
fig.show()
```

```python
import plotly.express as px


indicador_x = df_normalizados.columns[3]  
indicador_y = df_normalizados.columns[4] 

fig = px.scatter(
    df_normalizados,
    x=indicador_x,
    y=indicador_y,
    hover_name=df_normalizados.columns[0],  
    color="Mean",  
    size="Mean",
    color_continuous_scale="Viridis",   
    title=f"Comparación entre {indicador_x} y {indicador_y} por País"
)
fig.show()
```
