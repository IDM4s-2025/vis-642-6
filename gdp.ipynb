```python
dfn=pd.read_csv("indicadores_normalizados.csv")
dfn.head()
```

```python
dfn.columns
```

```python
dfn['Housing'] = dfn[['HO_HISH', 'HO_NUMR', 'HS_LEB']].mean(axis=1)
dfn['Income'] = dfn[['IW_HADI', 'IW_HNFW']].mean(axis=1)
dfn['Jobs'] = dfn[['JE_EMPL', 'JE_LMIS', 'JE_LTUR','JE_PEARN']].mean(axis=1)
dfn['Community'] = dfn[['SC_SNTWS']].mean(axis=1)
dfn['Education'] = dfn[['ES_EDUA','ES_EDUEX', 'ES_STCS']].mean(axis=1)
dfn['Enviroment'] = dfn[['EQ_AIRP', 'EQ_WATER']].mean(axis=1)
dfn['Civic engagement'] = dfn[['CG_SENG', 'CG_VOTO']].mean(axis=1)
dfn['Health'] = dfn[['HS_LEB','HS_SFRH']].mean(axis=1)
dfn['Life Satisfaction'] = dfn[['SW_LIFS']].mean(axis=1)
dfn['Safety'] = dfn[['PS_FSAFEN', 'PS_REPH']].mean(axis=1)
dfn['Work Life Balance'] = dfn[['WL_EWLH','WL_TNOW']].mean(axis=1)
df_final=dfn[['LOCATION','Housing','Income','Community','Education','Enviroment','Civic engagement','Health','Life Satisfaction','Safety','Work Life Balance','ny_gdp_pcap_pp_kd']]
df_final.head()
```

```python
df_final=pd.merge(df_final,df,on='LOCATION')
df_final.head()
```

```python
df_final['Promedio General']=df_final.iloc[:,1:11].mean(axis=1)
df_final.head()
```

```python
america = ['BRA', 'CAN', 'CHL', 'COL', 'CRI', 'MEX', 'USA']
europa = ['AUT', 'BEL', 'CHE', 'CZE', 'DEU', 'DNK', 'ESP', 'EST', 'FIN', 'FRA', 'GBR', 'GRC', 'HUN', 'IRL', 'ISL', 'ITA', 'LTU', 'LUX', 'LVA', 'NLD', 'NOR', 'POL', 'PRT', 'RUS', 'SVK', 'SVN', 'SWE']
asia = ['ISR', 'JPN', 'KOR', 'TUR']
oceania = ['AUS', 'NZL']
africa = ['ZAF']
df_final["Continente"] = None
for index, x in enumerate(df_final["LOCATION"]): 
    if x in america:
        df_final.at[index, "Continente"] = "America" 
    elif x in europa:
        df_final.at[index, "Continente"] = "Europa"
    elif x in asia:
        df_final.at[index, "Continente"] = "Asia"
    elif x in oceania:
        df_final.at[index, "Continente"] = "Oceania"
    elif x in africa:
        df_final.at[index, "Continente"] = "Africa"
```

```python
import plotly.express as px


fig = px.scatter(df_final,x='GDP',y='Promedio General',color='Continente',hover_name='LOCATION',size_max=60, title="Índice de bienestar vs PIB per cápita", trendline="ols", trendline_color_override="black", trendline_scope="overall")
fig.show()
```
